# Handling Missing Data

```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Aarav", "Sita", "Rahul", "Priya", "Samir", "Anisha", "Rohan", "Maya", "Kiran", "Nisha"],
    "Age": [24,  np.nan, 29, 32,  np.nan, 27, 35,  np.nan, 31, 26],
    "Department": ["IT", "HR", np.nan, "Finance", "IT", "Marketing", "HR", np.nan, "IT", "Finance"],
    "Salary": [45000, 52000, np.nan, 68000, 55000, np.nan, 72000, 48000, np.nan, 51000],
    "Experience": [1, 3, 5, np.nan, 4, 2, 8, 1, np.nan, 3]
}

df = pd.DataFrame(data)
df
```

# Finding Missing Data

```python
# Find where the value is NaN
df.isna() # True indicates that the vaue is NaN
```

```python
# Find which which column contains NaN values
df.isna().any()
```

```python
# Find the total number of NaN values in each column
df.isna().sum()
```

# Removing Missing Data

```python
df
```

```python
# Removing all the rows that has NaN values
df.dropna()
```

```python
# Removing rows with atleast 'x' number of non-NaN values
df.dropna(thresh=5) # The output will be only the rows that has at lesat 5 non-NaN values
```

# Filling Missing Values

```python
df
```

```python
# Filling all NaN values with a specific data
df.fillna(0) # This doesn't changes the original DataFrame. To Change the original one we must wirte 'inplace=True',
```

```python
# Filling column specific NaN values
fill_values = {"Age":20, "Department":"IT", "Salary":50000, "Experience":1}
df.fillna(fill_values) # Use 'inplace=True' to change the original DataFrame
```

```python
df["Age"].fillna(df["Age"].mean())
```
