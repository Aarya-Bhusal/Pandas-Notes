# GroupBy and Aggregation

```python
import numpy as np
import pandas as pd
```

# GroupBy

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
df
```

```python
# Groupping by Category
catergory = df.groupby("Category")
for cat1, cat2 in catergory:
    print(cat1)
    print(cat2)
```

```python
# Group by Category and calculate the sum of Sales
catergory = df.groupby("Category")["Sales"].sum()
catergory
```

```python
# Group by Store and calculate the sum of Sales
stores = df.groupby("Store")["Sales"].sum()
stores
```

```python
# Group by multiple columns
cat_and_stores = df.groupby(["Category", "Store"])["Sales"].sum()
cat_and_stores
```
