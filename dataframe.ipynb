# DataFrames in Pandas

```python
import numpy as np
import pandas as pd
```

# Creating a DataFrame

```python
# Using a Dict
data = {
    "Name":["Aarya","Aayush","Samar","Sachin"],
    "Age":[14, 13, 12, 15],
    "Class":[9, 8, 7, 10]
}
df = pd.DataFrame(data)
df
```

```python
# Using a List
data = [
    ["Aarya", 14, 9],
    ["Aayush", 13, 8],
    ["Samar", 12, 7],
    ["Sachin", 15, 10]
]
col = ["Name","Age","Grade"]
df = pd.DataFrame(data, columns=col)
df
```

# Selecting and Indexing of Columns

```python
data = {
    "Name":["Aarya","Aayush","Samar","Sachin"],
    "Age":[14, 13, 12, 15],
    "Class":[9, 8, 7, 10]
}
df = pd.DataFrame(data)
df
```

```python
# Selecting Single Column
df["Name"]
```

```python
# Selecting Multiple Column
df[["Name", "Class"]]
```

# Creating a new column

```python
df["GPA"] = [4,3.9,3.2,2] # The number of elements in the new column must be equal to the  number of elements in   
                          # the old columns.
df
```

# Removing a column

```python
df.drop(['Age','GPA'], axis=1) # This just returns a copy by removing the given column
```

```python
# To remove the column from the original DataFrame
df.drop(['Age', 'GPA'], axis=1, inplace=True)
df
```

# Selecting a Row

```python
data = {
    "Name": ["Aarya", "Rahul", "Priya", "Suman", "Rohan", "Anisha", "Kiran", "Sneha", "Aayush", "Nisha"],
    "Age": [22, 28, 25, 32, 24, 29, 35, 26, 23, 31],
    "Company": ["Google", "Microsoft", "Amazon", "Google", "Meta", "Amazon", "Microsoft", "Apple", "Meta", "Google"],
    "Salary": [75000, 95000, 82000, 110000, 78000, 90000, 120000, 85000, 72000, 105000],
    "Experience": [1, 5, 3, 8, 2, 6, 10, 4, 1, 7]
}
df = pd.DataFrame(data)
df
```

```python
# Selecting a single row
df.loc[0] # Output is a series
```

```python
# Selecting multiple rows
df.loc[[0,2]] # Output is a DataFrame
```

# Selecting Subsets of Rows and Columns

```python
df
```

```python
# Selecting two name and %
df.loc[[2,3]][["Name", "Company", "Salary"]]
```

```python
# Selecting multiple rows
df.iloc[2:9][["Name", "Company", "Salary"]]
```

# Conditional Selection

```python
# Selecting Rows where Age is >30
mask = df["Age"]>30
df[mask]
```

```python
# Selecting Rows where Age is >30 and Experience is >7.
mask = (df["Age"]>30) & (df["Experience"]>7)
df[mask]
```
