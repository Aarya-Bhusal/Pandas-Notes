# DataFrames in Pandas

```python
import numpy as np
import pandas as pd
```

# Creating a DataFrame

```python
# Using a Dict
data = {
    "Name":["Aarya","Aayush","Samar","Sachin"],
    "Age":[14, 13, 12, 15],
    "Class":[9, 8, 7, 10]
}
df = pd.DataFrame(data)
df
```

```python
# Using a List
data = [
    ["Aarya", 14, 9],
    ["Aayush", 13, 8],
    ["Samar", 12, 7],
    ["Sachin", 15, 10]
]
col = ["Name","Age","Grade"]
df = pd.DataFrame(data, columns=col)
df
```

# Selecting and Indexing of Columns

```python
data = {
    "Name":["Aarya","Aayush","Samar","Sachin"],
    "Age":[14, 13, 12, 15],
    "Class":[9, 8, 7, 10]
}
df = pd.DataFrame(data)
df
```

```python
# Selecting Single Column
df["Name"]
```

```python
# Selecting Multiple Column
df[["Name", "Class"]]
```

# Creating a new column

```python
df["GPA"] = [4,3.9,3.2,2] # The number of elements in the new column must be equal to the  number of elements in   
                          # the old columns.
df
```

# Removing a column

```python
df.drop(['Age','GPA'], axis=1) # This just returns a copy by removing the given column
```

```python
# To remove the column from the original DataFrame
df.drop(['Age', 'GPA'], axis=1, inplace=True)
df
```
