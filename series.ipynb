# Series in Pandas
**Series is a 1-D labelled array capable of holding any data types.**

```python
import numpy as np
import pandas as pd
```

```python
# Creating Series using a lists
label_list = ["Name", "Age", "Grade"]
data_list = ["Aarya", 14, 9]
series = pd.Series(data_list, label_list)
series
```

```python
# Creating Series using a Array
label_list = np.array(["Name", "Age", "Grade"])
data_list = np.array(["Aarya", "14", "9"])
series = pd.Series(data_list, label_list)
series
```

```python
# Creating Series using a Array
data_dict = {
    "Name":"Aarya",
    "Age":14,
    "Grade":9
}
series = pd.Series(data_dict)
series
```
